```python
import pandas as pd
import numpy as np
import sklearn as sk
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
# Reading dataset
unknowndataset = pd.read_csv("Assignment3-Unknown-Dataset.csv")
unknowndataset
```

```python
# Total number of columns
print(len(unknowndataset.columns))
```

# Data preprocessing 

```python
#Check missing values in each columns 
missing_values = unknowndataset.isin(['?', 'unknown']).sum()
print(missing_values)
```

## Clean data
- Keep all columns

```python
# Impute other attributes
for column in unknowndataset:
    # Replace missing values with pandas NaN null then fill
    unknowndataset[column] = unknowndataset[column].replace(['?', 'unknown'], pd.NA)
    # Check if there are missing values null
    if unknowndataset[column].isnull().sum() > 0:
        if unknowndataset[column].dtype == 'object':
            fill_mode = unknowndataset[column].mode()[0] # incase many modes, get the first one
            unknowndataset[column].fillna(fill_mode, inplace=True)
        else:
            # Numeric column → impute with mean
            fill_mean = unknowndataset[column].mean()
            unknowndataset[column].fillna(fill_mean, inplace=True)
```

```python
unknowndataset
```

```python
# Check again if there are any missing values
any_missing_values = unknowndataset.isin(['?', 'unknown']).sum()
any_missing_values
```

```python
# Check unique values
for col in unknowndataset.columns:
    print(f"Column: {col}")
    print(unknowndataset[col].unique())
```

# Normalization
- Using minmax scaling
- For duration, pdays,emp.var.rate	cons.price.idx	cons.conf.idx, euribor3m

```python
from sklearn.preprocessing import MinMaxScaler

# Scaler
min_max_scaler = MinMaxScaler()

# Chosen columns to normalise
col_to_normalise = [    'age', 'duration','emp.var.rate','cons.price.idx', 'cons.conf.idx', 
    'euribor3m', 'nr.employed', 'pdays']

# Normalize the columns
for col in col_to_normalise:
    unknowndataset[col] = min_max_scaler.fit_transform(unknowndataset[col].values.reshape(-1, 1))
```

# Data transformation
- Encode job, marital, education, contact, month, day_of_week, campaign, poutcome, state, subscribed? 
- Encode binary columns: deafult, housing, loan, previous 

```python
from sklearn.preprocessing import LabelEncoder, OneHotEncoder

# Initialize the LabelEncoder
label_encoder = LabelEncoder()
label_encode_columns = ['job', 'marital', 'education', 'campaign', 'month', 'day_of_week', 'state']

# Apply LabelEncoder to each column
for column in label_encode_columns:
    unknowndataset[column] = label_encoder.fit_transform(unknowndataset[column].values.reshape(-1, 1))

# Initialize the OneHoteEncoder
onehot_encoder = OneHotEncoder(dtype=int, sparse_output=False)
onehot_encode_columns = ['contact', 'default', 'housing', 'loan', 'poutcome', 'previous']
for col in onehot_encode_columns:
        unknowndataset[col] = onehot_encoder.fit_transform(unknowndataset[col].values.reshape(-1, 1))
```

# Feature selection
- Similar to marketing dataset
- Filter relevant columns

```python
sub_unknowndataset = unknowndataset[['row ID', 'duration', 'campaign', 'nr.employed', 'euribor3m', 'emp.var.rate', 'cons.price.idx','previous', 'marital', 'education']]
```

```python
sub_unknowndataset
```

```python
X_unknown = sub_unknowndataset.drop(columns=['row ID'])
```

# Prediction

```python
import joblib

# Load the trained models
dt_model = joblib.load('dt_model.pkl')
rf_model = joblib.load('rf_model.pkl')
knn_model = joblib.load('knn_model.pkl')
ada_model = joblib.load('ada_model.pkl')
nn_model = joblib.load('nn_model.pkl')
nusvc_model = joblib.load('nusvc_model.pkl')
```

```python
# Prediction
y_pred_dt = dt_model.predict(X_unknown)
y_pred_rf = rf_model.predict(X_unknown)
y_pred_knn = knn_model.predict(X_unknown)
y_pred_ada = ada_model.predict(X_unknown)
y_pred_nn = nn_model.predict(X_unknown)
y_pred_nusvc = nusvc_model.predict(X_unknown)
```

```python
# Output csv
# For decision tree
dt_output = sub_unknowndataset[['row ID']]
dt_output['subscribed'] = y_pred_dt
dt_output.to_csv('dt_predictions.csv', index=False)

# For random forest
rf_output = sub_unknowndataset[['row ID']]
rf_output['subscribed'] = y_pred_rf
rf_output.to_csv('rf_predictions.csv', index=False)

# For knn
knn_output = sub_unknowndataset[['row ID']]
knn_output['subscribed'] = y_pred_knn
knn_output.to_csv('knn_predictions.csv', index=False)

# For ada
ada_output = sub_unknowndataset[['row ID']]
ada_output['subscribed'] = y_pred_ada
ada_output.to_csv('ada_predictions.csv', index=False)

# For nn
nn_output = sub_unknowndataset[['row ID']]
nn_output['subscribed'] = y_pred_nn
nn_output.to_csv('nn_predictions.csv', index=False)

# For svm
nusvc_output = sub_unknowndataset[['row ID']]
nusvc_output['subscribed'] = y_pred_nusvc
nusvc_output.to_csv('nusvc_predictions.csv', index=False)

#Verfiy
print("All saved")
```
